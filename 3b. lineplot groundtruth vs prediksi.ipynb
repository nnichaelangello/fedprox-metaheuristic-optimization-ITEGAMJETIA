```python
import json
import time
import psutil
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from scipy.stats import f_oneway
from mealpy import FloatVar
from mealpy.swarm_based import PSO, GWO, HHO
from mealpy.evolutionary_based import DE, GA
import matplotlib.pyplot as plt

def load_and_prepare_data():
    url = "https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/insurance.csv"
    df = pd.read_csv(url)
    return df

def check_heterogeneity(clients_data):
    groups = [client['y'] for client in clients_data]
    f_stat, p_value = f_oneway(*groups)
    f_stat = float(f_stat)
    p_value = float(p_value)
    print(f"ANOVA test for heterogeneity: F = {f_stat:.4f}, p = {p_value:.4e}")
    print("Clients are significantly heterogeneous:", "True" if p_value < 0.05 else "False")

def split_into_clients(df, n_clients=5):
    df_grouped = df.groupby(['region', 'smoker'])
    client_dfs = [group for _, group in df_grouped]
    while len(client_dfs) < n_clients:
        largest = max(client_dfs, key=len)
        client_dfs.remove(largest)
        split_idx = len(largest) // 2
        client_dfs.append(largest.iloc[:split_idx])
        client_dfs.append(largest.iloc[split_idx:])
    clients = []
    for cdf in client_dfs[:n_clients]:
        clients.append({'raw': cdf})
    check_heterogeneity([{'y': c['raw']['charges'].values} for c in clients])
    return clients

class RegressionNet(nn.Module):
    def __init__(self, input_dim):
        super(RegressionNet, self).__init__()
        self.fc1 = nn.Linear(input_dim, 64)
        self.fc2 = nn.Linear(64, 32)
        self.fc3 = nn.Linear(32, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)
        return x

def train_local_model(model, X_train, y_train, device, epochs=10, lr=0.001):
    model.train()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()
    dataset = TensorDataset(torch.FloatTensor(X_train), torch.FloatTensor(y_train))
    loader = DataLoader(dataset, batch_size=32, shuffle=True)
    for _ in range(epochs):
        for batch_x, batch_y in loader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)
            optimizer.zero_grad()
            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()

def fedprox_aggregate(global_model, client_models, mu, device):
    global_dict = global_model.state_dict()
    for key in global_dict.keys():
        if 'weight' in key or 'bias' in key:
            other_params = [cm.state_dict()[key].float() for cm in client_models]
            avg = torch.mean(torch.stack(other_params), dim=0)
            proximal_diff = mu * (global_dict[key] - avg)
            global_dict[key] = avg + proximal_diff
    global_model.load_state_dict(global_dict)

def get_method_optimizer(method_name, problem_dict):
    if method_name == "DE":
        return DE.OriginalDE(epoch=100, pop_size=50, **problem_dict)
    elif method_name == "PSO":
        return PSO.OriginalPSO(epoch=100, pop_size=50, **problem_dict)
    elif method_name == "GWO":
        return GWO.OriginalGWO(epoch=100, pop_size=50, **problem_dict)
    elif method_name == "HHO":
        return HHO.OriginalHHO(epoch=100, pop_size=50, **problem_dict)
    elif method_name == "GA":
        return GA.OriginalGA(epoch=100, pop_size=50, **problem_dict)
    return None

def tune_proximal_param(method_name, client_models, global_model, device, input_dim, clients):
    start_time = time.time()
    process = psutil.Process()
    start_mem = process.memory_info().rss / (1024 ** 2)

    def objective(mu_vec):
        mu = mu_vec[0]
        temp_global = RegressionNet(input_dim).to(device)
        temp_global.load_state_dict(global_model.state_dict())
        fedprox_aggregate(temp_global, client_models, mu, device)
        total_loss = 0.0
        for i, client in enumerate(clients):
            X_test = client['X_test']
            y_test_scaled = client['y_test']
            temp_global.eval()
            with torch.no_grad():
                pred_scaled = temp_global(torch.FloatTensor(X_test).to(device)).cpu().numpy().flatten()
            pred = client['scaler_y'].inverse_transform(pred_scaled.reshape(-1, 1)).flatten()
            actual = client['scaler_y'].inverse_transform(y_test_scaled).flatten()
            total_loss += np.mean((pred - actual) ** 2)
        return total_loss / len(clients)

    problem_dict = {
        "obj_func": objective,
        "bounds": FloatVar(lb=[0.001], ub=[1.0]),
        "minmax": "min",
        "verbose": False
    }

    if method_name in ["Grid Search", "Random Search"]:
        candidates = np.linspace(0.001, 1.0, 40)
        if method_name == "Random Search":
            np.random.shuffle(candidates)
        best_mu = 0.001
        best_score = float('inf')
        for mu in candidates[:30]:
            score = objective([mu])
            if score < best_score:
                best_score = score
                best_mu = mu
        mu_best = best_mu
    else:
        optimizer = get_method_optimizer(method_name, problem_dict)
        if optimizer is None:
            raise ValueError(f"Unsupported method: {method_name}")
        best_agent = optimizer.solve(problem=problem_dict)
        mu_best = best_agent.solution[0]

    end_time = time.time()
    end_mem = process.memory_info().rss / (1024 ** 2)
    search_time = end_time - start_time
    total_memory = max(0.0, end_mem - start_mem)

    return mu_best, search_time, total_memory

def prepare_client_data(client):
    df_c = client['raw']
    X = df_c.drop('charges', axis=1)
    y = df_c['charges'].values.reshape(-1, 1)

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), ['age', 'bmi', 'children']),
            ('cat', OneHotEncoder(drop='first', sparse_output=False), ['sex', 'smoker', 'region'])
        ])

    n_samples = len(X)
    train_size = int(0.8 * n_samples)
    X_train_raw = X.iloc[:train_size]
    X_test_raw  = X.iloc[train_size:]
    y_train_raw = y[:train_size]
    y_test_raw  = y[train_size:]

    X_train = preprocessor.fit_transform(X_train_raw)
    X_test  = preprocessor.transform(X_test_raw)

    scaler_y = StandardScaler()
    y_train_scaled = scaler_y.fit_transform(y_train_raw)
    y_test_scaled  = scaler_y.transform(y_test_raw)

    return {
        'X_train': X_train,
        'X_test': X_test,
        'y_train': y_train_scaled,
        'y_test': y_test_scaled,
        'scaler_y': scaler_y,
        'preprocessor': preprocessor
    }

def run_experiment():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    df_raw = load_and_prepare_data()
    clients_raw = split_into_clients(df_raw)

    clients = []
    for c in clients_raw:
        processed = prepare_client_data(c)
        clients.append(processed)

    input_dim = clients[0]['X_train'].shape[1]
    methods = ["DE", "PSO", "GWO", "HHO", "GA", "Grid Search", "Random Search"]

    global_models = {m: RegressionNet(input_dim).to(device) for m in methods}

    print("Starting Federated Learning with FedProx - 50 rounds")
    print("No performance metrics will be computed or displayed.\n")

    for round_num in range(1, 51):
        print(f"Round {round_num}/50")

        client_models_per_method = {m: [] for m in methods}

        for client in clients:
            X_train = client['X_train']
            y_train = client['y_train']

            for m in methods:
                model = RegressionNet(input_dim).to(device)
                if round_num > 1:
                    model.load_state_dict(global_models[m].state_dict())
                train_local_model(model, X_train, y_train, device)
                client_models_per_method[m].append(model)

        for m in methods:
            mu, _, _ = tune_proximal_param(
                m, client_models_per_method[m], global_models[m], device, input_dim, clients
            )
            fedprox_aggregate(global_models[m], client_models_per_method[m], mu, device)

    print("\nAll 50 rounds completed. Generating comparison plot...")

    all_actual = []
    all_predictions = {m: [] for m in methods}

    for client in clients:
        X_test = client['X_test']
        y_test_scaled = client['y_test']
        scaler_y = client['scaler_y']

        actual = scaler_y.inverse_transform(y_test_scaled).flatten()
        all_actual.extend(actual)

        for m in methods:
            global_models[m].eval()
            with torch.no_grad():
                pred_scaled = global_models[m](torch.FloatTensor(X_test).to(device)).cpu().numpy().flatten()
            pred = scaler_y.inverse_transform(pred_scaled.reshape(-1, 1)).flatten()
            all_predictions[m].extend(pred)

    all_actual = np.array(all_actual)
    index = np.arange(1, len(all_actual) + 1)

    plt.style.use('seaborn-v0_8-whitegrid')
    fig, ax = plt.subplots(figsize=(12, 7), dpi=120)

    colors = {
        "Ground Truth": "black",
        "DE": "#1f77b4",
        "PSO": "#ff7f0e",
        "GWO": "#2ca02c",
        "HHO": "#d62728",
        "GA": "#9467bd",
        "Grid Search": "#8c564b",
        "Random Search": "#e377c2"
    }

    ax.plot(index, all_actual, label="Ground Truth", color=colors["Ground Truth"], linewidth=2.8, linestyle='-', alpha=0.9)

    for m in methods:
        ax.plot(index, all_predictions[m], label=m, color=colors[m], linewidth=1.6, alpha=0.75)

    ax.set_xlabel("Test Sample Index", fontsize=14, fontweight='medium')
    ax.set_ylabel("Insurance Charges ($)", fontsize=14, fontweight='medium')
    ax.set_title("Global Model Predictions vs Ground Truth after 50 Rounds (FedProx)", fontsize=16, fontweight='bold', pad=15)

    ax.legend(fontsize=11, loc='upper left', frameon=True, edgecolor='gray', ncol=2)
    ax.tick_params(axis='both', which='major', labelsize=11)
    ax.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    run_experiment()
```
